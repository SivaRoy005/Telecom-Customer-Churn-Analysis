# Customer Churn Analysis: Data-Driven Insights
**Objective:** To analyze customer data and identify the key factors causing customers to leave (churn). 
**Dataset:** Telecom Customer Churn Data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set(style="whitegrid")
```

```python
# Load the dataset
df = pd.read_csv('Customer Churn dataset.csv')

# Check first 5 rows
print(df.head())

# Check data types and missing values
print(df.info())
```

```python
# Convert TotalCharges to numeric, setting errors to 'coerce' will turn empty spaces into NaN
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Drop the rows with missing values (usually very few)
df.dropna(inplace=True)

# Change Churn from 'Yes/No' to 1/0 for better analysis if needed
df['Churn_Numeric'] = df['Churn'].apply(lambda x: 1 if x == 'Yes' else 0)
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Churn', data=df, hue='Churn', palette='viridis', legend=False)
plt.title('Overall Churn Distribution')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x='Contract', hue='Churn', data=df, palette='magma')
plt.title('Churn Rate by Contract Type')
plt.show()
```

### Key Observation:
* **Contract Risk:** As seen in the graph, customers on **Month-to-month** contracts have a significantly higher churn rate compared to those on one or two-year contracts.
* **Strategy:** The business should focus on converting short-term customers into long-term contracts through loyalty discounts.

```python
plt.figure(figsize=(12, 8))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Customer Metrics')
plt.show()
```

## Final Summary
1. **High Risk Segment:** Month-to-month contract users and Fiber Optic users.
2. **Recommendation 1:** Introduce a "contract upgrade" campaign.
3. **Recommendation 2:** Investigate service quality issues for Fiber Optic customers.
